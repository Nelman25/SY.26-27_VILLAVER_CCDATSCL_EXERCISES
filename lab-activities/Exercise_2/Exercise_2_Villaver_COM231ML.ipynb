**26. Connect and create a table (5 pts)**

- (a) Import `sqlite3` and connect to a database named `'school.db'`. Save the connection in a variable named `conn` and print `"Opened database successfully"`.
- (b) Run this line first so the cell can be re-run without errors (it deletes the old table if it already exists):
  ```python
  conn.execute('DROP TABLE IF EXISTS STUDENTS')
  ```
- (c) Create a table named `STUDENTS` with these columns, then print `"Table created successfully"`:

| Column | Type | Rule |
|---|---|---|
| `ID` | `INT` | `PRIMARY KEY`, `NOT NULL` |
| `NAME` | `TEXT` | `NOT NULL` |
| `AGE` | `INT` | `NOT NULL` |
| `COURSE` | `CHAR(50)` | |
| `GWA` | `REAL` | |

````python
# Write your code here
import sqlite3

conn = sqlite3.connect('school.db')
print('Opened database successfully')

conn.execute('DROP TABLE IF EXISTS STUDENTS')

conn.execute('''
    CREATE TABLE STUDENTS (
        ID INT PRIMARY KEY NOT NULL,
        NAME TEXT NOT NULL,
        AGE INT NOT NULL,
        COURSE CHAR(50),
        GWA REAL
    )
''')
conn.commit()

print("Table created successfully")
````

**28. Update and delete records (5 pts)**

- (a) **Update** Carla's record (`ID = 3`) so that her `GWA` becomes `1.25`. Commit the change.
- (b) **Delete** Dan's record (`ID = 4`). Commit the change.
- (c) Print `conn.total_changes`.
- (d) Select `ID`, `NAME` and `GWA` from `STUDENTS` and print every remaining record using a `for` loop.
- (e) Close the connection.

````python
# Write your code here
conn.execute('''
    UPDATE STUDENTS
    SET GWA = 1.25
    WHERE ID = 3
''')
conn.commit()

conn.execute('''
    DELETE FROM STUDENTS
    WHERE ID = 4
''')
conn.commit()

print(f"Total changes: {conn.total_changes} \n")

cursor = conn.execute('''
    SELECT ID, NAME, GWA
    FROM STUDENTS
''')

for row in cursor:
    print(f"ID = {row[0]}")
    print(f"NAME = {row[1]}")
    print(f"GWA = {row[2]}")
    print()

conn.close()
````
